```python
from pathlib import Path

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import torch
from IPython.display import display
from medmnist import INFO, PathMNIST
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, confusion_matrix, f1_score
from torch import nn
from torch.utils.data import DataLoader, TensorDataset

SEED = 42
TRAIN_SIZE = 10_000
BATCH_SIZE = 128
EPOCHS = 5

rng = np.random.default_rng(SEED)
torch.manual_seed(SEED)
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
if device.type == "cpu":
    torch.set_num_threads(min(4, torch.get_num_threads()))
print(f"Using {device}")
```

```python
data_dir = Path("data")
data_dir.mkdir(exist_ok=True)

train = PathMNIST(split="train", root=str(data_dir), download=True)
val = PathMNIST(split="val", root=str(data_dir), download=True)

class_ids = sorted(int(label) for label in INFO["pathmnist"]["label"])
class_names = [INFO["pathmnist"]["label"][str(label)] for label in class_ids]
short_names = [
    "Adipose", "Background", "Debris", "Lymphocytes", "Mucus",
    "Smooth muscle", "Normal mucosa", "Cancer stroma", "Adenocarcinoma",
]

train_indices = rng.choice(len(train), TRAIN_SIZE, replace=False)
train_images = train.imgs[train_indices]
train_labels = train.labels.ravel()[train_indices]
val_labels = val.labels.ravel()
print(f"Training subset has {len(train_images):,} images. Validation has {len(val):,}.")

fig, axes = plt.subplots(3, 3, figsize=(7, 7), layout="constrained")
for label, ax in zip(class_ids, axes.flat):
    example_index = np.flatnonzero(train_labels == label)[0]
    ax.imshow(train_images[example_index])
    ax.set_title(short_names[label])
    ax.axis("off")
plt.show()
```

```python
def mean_rgb(images):
    return images.mean(axis=(1, 2)) / 255.0


def scores(labels, predictions):
    return {
        "Accuracy": accuracy_score(labels, predictions),
        "Macro F1": f1_score(
            labels, predictions, labels=class_ids, average="macro", zero_division=0
        ),
    }


baseline = LogisticRegression(max_iter=500)
baseline.fit(mean_rgb(train_images), train_labels)
baseline_val = scores(val_labels, baseline.predict(mean_rgb(val.imgs)))
print(f"Validation accuracy: {baseline_val['Accuracy']:.2%}")
print(f"Validation macro F1: {baseline_val['Macro F1']:.3f}")
```

```python
def image_dataset(images, labels):
    # PyTorch convolutions expect images in N × channels × height × width order.
    pixels = torch.from_numpy(images).permute(0, 3, 1, 2)
    pixels = pixels.float().div_(255.0)
    targets = torch.as_tensor(labels, dtype=torch.long)
    return TensorDataset(pixels, targets)


train_loader = DataLoader(
    image_dataset(train_images, train_labels), batch_size=BATCH_SIZE,
    shuffle=True, generator=torch.Generator().manual_seed(SEED),
)
val_loader = DataLoader(
    image_dataset(val.imgs, val_labels), batch_size=BATCH_SIZE,
)

torch.manual_seed(SEED)
cnn = nn.Sequential(
    nn.Conv2d(3, 16, kernel_size=3, padding=1),
    nn.ReLU(),
    nn.MaxPool2d(2),
    nn.Conv2d(16, 32, kernel_size=3, padding=1),
    nn.ReLU(),
    nn.MaxPool2d(2),
    nn.Flatten(),
    nn.Linear(32 * 7 * 7, len(class_ids)),
).to(device)
optimizer = torch.optim.Adam(cnn.parameters(), lr=0.001)
loss_fn = nn.CrossEntropyLoss()


@torch.inference_mode()
def predict_cnn(loader):
    cnn.eval()
    predictions = []
    for images, _ in loader:
        logits = cnn(images.to(device))
        predictions.append(logits.argmax(dim=1).cpu())
    return torch.cat(predictions).numpy()


for epoch in range(EPOCHS):
    cnn.train()
    total_loss = 0.0
    for images, labels in train_loader:
        images, labels = images.to(device), labels.to(device)
        optimizer.zero_grad()
        loss = loss_fn(cnn(images), labels)
        loss.backward()
        optimizer.step()
        total_loss += loss.item() * len(labels)

    cnn_val = scores(val_labels, predict_cnn(val_loader))
    mean_loss = total_loss / len(train_loader.dataset)
    print(
        f"Epoch {epoch + 1}/{EPOCHS} | Train loss {mean_loss:.3f} | "
        f"Val accuracy {cnn_val['Accuracy']:.2%} | Val macro F1 {cnn_val['Macro F1']:.3f}"
    )
```

```python
test = PathMNIST(split="test", root=str(data_dir), download=True)
test_labels = test.labels.ravel()
test_loader = DataLoader(
    image_dataset(test.imgs, test_labels), batch_size=BATCH_SIZE,
)

baseline_test = scores(test_labels, baseline.predict(mean_rgb(test.imgs)))
cnn_test_predictions = predict_cnn(test_loader)
cnn_test = scores(test_labels, cnn_test_predictions)

results = pd.DataFrame([
    {"Model": "Mean RGB baseline", "Split": "Validation", **baseline_val},
    {"Model": "Two-layer CNN", "Split": "Validation", **cnn_val},
    {"Model": "Mean RGB baseline", "Split": "Test", **baseline_test},
    {"Model": "Two-layer CNN", "Split": "Test", **cnn_test},
])

display(results.style.format({"Accuracy": "{:.2%}", "Macro F1": "{:.3f}"}))
```

```python
matrix = confusion_matrix(
    test_labels, cnn_test_predictions, labels=class_ids, normalize="true"
)
fig, ax = plt.subplots(figsize=(9, 7), layout="constrained")
heatmap = ax.imshow(matrix, cmap="Blues", vmin=0, vmax=1)
fig.colorbar(heatmap, ax=ax, label="Fraction of true class")
ax.set_xticks(class_ids, short_names, rotation=45, ha="right")
ax.set_yticks(class_ids, short_names)
ax.set(xlabel="Predicted class", ylabel="True class", title="CNN test confusion matrix")
for row in class_ids:
    for column in class_ids:
        value = matrix[row, column]
        if value >= 0.005:
            color = "white" if value > 0.5 else "black"
            ax.text(column, row, f"{value:.0%}", ha="center", va="center", color=color)
plt.show()

errors = matrix.copy()
np.fill_diagonal(errors, 0)
true_class, predicted_class = np.unravel_index(errors.argmax(), errors.shape)
if errors[true_class, predicted_class] > 0:
    print(
        f"{errors[true_class, predicted_class]:.1%} of test patches labeled "
        f"{class_names[true_class]} were predicted as {class_names[predicted_class]}."
    )
    example_indices = np.flatnonzero(
        (test_labels == true_class) & (cnn_test_predictions == predicted_class)
    )[:3]
    fig, axes = plt.subplots(1, len(example_indices), figsize=(7, 2.5), squeeze=False)
    for ax, example_index in zip(axes.flat, example_indices):
        ax.imshow(test.imgs[example_index])
        ax.axis("off")
    plt.show()
```
